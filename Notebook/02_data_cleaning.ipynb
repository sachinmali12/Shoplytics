```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Data/SuperStore_Sales_Dataset.csv")
df
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst=True)

df
```

```python
# Handle Missing Values

df['Returns'] = df['Returns'].fillna('No')
df
```

```python
#  Drop unnecessary columns (ind1, ind2)



df.drop(columns=['ind1', 'ind2'], errors='ignore', inplace=True)
df.columns
```

```python
 # Remove duplicate rows from dataset


df.drop_duplicates(inplace=True)
```

```python
# TASK 7: Clean and standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(' ', '_')
)

df.info()
```

```python
# TASK 9: Save the cleaned dataset for further analysis
df.to_csv("data/superstore_cleaned.csv", index=False)
```
